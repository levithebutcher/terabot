### 🛡️ Step 3: Ultra-Strong Anti-Disconnect & Keep-Alive System
Colab ko background me disconnect ya sleep hone se bachane ke liye is cell ko run karein.

Yeh cell **3 layers of protection** active karta hai:
1. 🔊 **Continuous Inaudible Audio Loop**: Chrome/Edge browser ko background tab sleep karne se rokta hai.
2. 🖱️ **Browser DOM Keep-Alive**: Colab reconnect button aur window activity ko auto-ping karta hai.
3. 🐍 **Kernel Python Daemon**: Background thread kernel idle timeout prevent karta hai.

```python
# 🛡️ Ultra-Strong Colab Keep-Alive & Anti-Disconnect Engine
import threading
import time
from IPython.display import HTML, display

# 1. Background Python Daemon Heartbeat
def _background_heartbeat():
    while True:
        time.sleep(60)
        try:
            with open('/content/.keepalive', 'w') as f:
                f.write(str(time.time()))
        except Exception:
            pass

if not any(t.name == 'colab_heartbeat_daemon' for t in threading.enumerate()):
    _t = threading.Thread(target=_background_heartbeat, name='colab_heartbeat_daemon', daemon=True)
    _t.start()

# 2. Browser DOM Keep-Alive & Inaudible Continuous Web Audio Engine
keepalive_html = """
<div style="background: linear-gradient(135deg, #161823 0%, #1e2235 100%); border-radius: 14px; padding: 18px 24px; color: #fff; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; box-shadow: 0 4px 20px rgba(0,0,0,0.35); border: 1px solid rgba(0,255,136,0.2); margin: 10px 0;">
  <div style="display: flex; align-items: center; justify-content: space-between; margin-bottom: 12px;">
    <div style="display: flex; align-items: center; gap: 10px;">
      <span style="display: inline-block; width: 14px; height: 14px; background: #00ff88; border-radius: 50%; box-shadow: 0 0 12px #00ff88;"></span>
      <h3 style="margin: 0; color: #fff; font-size: 16px; font-weight: 600;">Ultra-Strong Keep-Alive Engine Active</h3>
    </div>
    <span style="background: rgba(0,255,136,0.15); color: #00ff88; font-size: 12px; font-weight: 700; padding: 4px 12px; border-radius: 20px; border: 1px solid rgba(0,255,136,0.3);">24/7 ACTIVE</span>
  </div>
  <div style="display: grid; grid-template-columns: repeat(auto-fit, minmax(200px, 1fr)); gap: 12px; font-size: 13px; color: #c4c8d8;">
    <div>🔊 <b>Audio Engine:</b> <span style="color: #00ff88;">Continuous 20Hz Loop</span></div>
    <div>🖱️ <b>DOM Activator:</b> <span style="color: #00ff88;">Auto-Ping (60s)</span></div>
    <div>🐍 <b>Kernel Heartbeat:</b> <span style="color: #00ff88;">Daemon Thread Running</span></div>
  </div>
</div>

<script>
// A. Inaudible Continuous Web Audio (Prevents tab throttling/sleep)
try {
  const AudioCtx = window.AudioContext || window.webkitAudioContext;
  if (AudioCtx) {
    const ctx = new AudioCtx();
    const osc = ctx.createOscillator();
    const gain = ctx.createGain();
    osc.type = 'sine';
    osc.frequency.setValueAtTime(20, ctx.currentTime); // Inaudible 20 Hz
    gain.gain.setValueAtTime(0.00001, ctx.currentTime); // Near-zero volume
    osc.connect(gain);
    gain.connect(ctx.destination);
    osc.start();
  }
} catch (e) {
  console.log('[KeepAlive] Audio Engine Note:', e);
}

// B. Browser DOM Auto-Ping
if (window._colabKeepAliveTimer) {
  clearInterval(window._colabKeepAliveTimer);
}
window._colabKeepAliveTimer = setInterval(() => {
  try {
    window.dispatchEvent(new MouseEvent('mousemove', { bubbles: true }));
    const btn = document.querySelector('colab-connect-button');
    if (btn && btn.shadowRoot) {
      const inner = btn.shadowRoot.querySelector('#connect');
      if (inner && inner.innerText && inner.innerText.toLowerCase().includes('reconnect')) {
        inner.click();
        console.log('[KeepAlive] Reconnected successfully.');
      }
    }
  } catch (err) {}
}, 60000);
</script>
"""

display(HTML(keepalive_html))
print('✅ Step 3 Complete: Ultra-Strong Keep-Alive system active!')
```

### 🚀 Step 4: Start Bot with Live Action Transparency Dashboard
Is cell ko play (▶️) karein. Yeh automatically:
1. 🔄 **GitHub se latest commits pull karega** (Restart karne par khud hi new code le lega).
2. 🧹 **Puraani stuck processes & database locks clear karega**.
3. 📊 **Pre-flight health dashboard dikhayega** (Token, Channel, MongoDB, Disk/RAM).
4. ⚡ **Bot actions ka live transparent feed dikhayega** (Connected, Incoming links, Mega Explorer, Cache skips, Downloads, Uploads).

```python
# 🚀 Step 4: High-Transparency Bot Runner with Auto-GitHub Update
import os
import sys
import time
import shutil
import subprocess

%cd /content/terabot

# 1. Auto-Sync with GitHub on every restart
print("=" * 65)
print("🔄 [1/4] AUTO-UPDATE: Checking GitHub for newest commits...")
print("=" * 65)
try:
    !git fetch origin main --quiet
    local_commit = !git rev-parse HEAD
    remote_commit = !git rev-parse origin/main
    if local_commit and remote_commit and local_commit[0] != remote_commit[0]:
        print(f"✨ New update found on GitHub! ({local_commit[0][:7]} -> {remote_commit[0][:7]})")
        !git reset --hard origin/main
        !pip install -q -r requirements.txt
        print("✅ Successfully updated to latest commit!")
    else:
        print("⚡ Already up-to-date with GitHub main branch.")
    !echo "📌 Active Commit: $(git log -1 --pretty=format:'%h - "%s" (%cr)')"
except Exception as e:
    print(f"⚠️ Git update notice: {e}")

# 2. Clean old stuck bot instances & locks
print("\n🧹 [2/4] CLEANUP: Clearing leftover processes and SQLite locks...")
!pkill -9 -f "python.*bot.py" 2>/dev/null || true
!rm -f terabot.db-journal terabot.db-wal downloads/*.part 2>/dev/null || true

# 3. Transparency Dashboard: Credentials & Environment Health Check
print("\n📊 [3/4] TRANSPARENCY DASHBOARD: Verifying System Health...")
if not os.path.exists('.env'):
    print("❌ ERROR: .env file missing! Please run Step 2 first.")
    raise FileNotFoundError(".env file not found")

env_vars = {}
with open('.env', 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if line and not line.startswith('#') and '=' in line:
            k, v = line.split('=', 1)
            env_vars[k.strip()] = v.strip()

token = env_vars.get('BOT_TOKEN', '')
masked_token = f"{token[:8]}...{token[-5:]}" if len(token) > 15 else "MISSING"
api_id = env_vars.get('API_ID', 'MISSING')
storage_ch = env_vars.get('PRIVATE_CHAT_ID', 'None')
has_mongo = bool(env_vars.get('MONGO_URI', ''))
has_cookie = bool(env_vars.get('TERABOX_COOKIE', ''))

total, used, free = shutil.disk_usage("/content")
free_gb = free // (2**30)
total_gb = total // (2**30)

print(f"   🤖 Bot Token:        {masked_token} {'✅' if len(token) > 15 else '❌'}")
print(f"   🔑 Telegram API ID:  {api_id} {'✅' if api_id != 'MISSING' else '❌'}")
print(f"   🛡️ Storage Channel:  {storage_ch} {'✅' if storage_ch else '⚠️ (DM Delivery Only)'}")
print(f"   ☁️ MongoDB Cloud:    {'Atlas Shared Cache Active' if has_mongo else 'Local SQLite Only'} {'✅' if has_mongo else '⚠️'}")
print(f"   🍪 TeraBox Cookie:   {'Provided' if has_cookie else 'None (Public Links)'} {'✅' if has_cookie else '⚠️'}")
print(f"   💾 Colab Disk Space:  {free_gb} GB free / {total_gb} GB total ✅")
print(f"   ⚡ Parallel Streams:  {env_vars.get('DOWNLOAD_STREAMS', '12')} workers ✅")

# 4. Launch Bot with Unbuffered Live Stream & Action Highlights
print("\n" + "=" * 65)
print("🚀 [4/4] LIVE BOT MONITORING CONSOLE")
print("=" * 65)
print("💡 Bot is launching... Real-time actions will appear below with tags.")
print("💡 Press Stop (⏹️) in Colab anytime to safely halt the bot.\n")

process = subprocess.Popen(
    [sys.executable, "-u", "bot.py"],
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    bufsize=1,
    universal_newlines=True,
    cwd="/content/terabot",
    env=dict(os.environ, PYTHONUNBUFFERED="1")
)

try:
    for line in iter(process.stdout.readline, ''):
        line = line.rstrip()
        if not line:
            continue
        
        # Action Transparency Highlights
        if "Bot successfully logged in as @" in line or "Bot is running as @" in line:
            print(f"\n🎉 [CONNECTED] {line}")
        elif "Routing Mega link" in line or "mega.nz" in line:
            print(f"📁 [MEGA ACTION] {line}")
        elif "Received download request" in line or "terabox" in line:
            print(f"📥 [TERABOX ACTION] {line}")
        elif "Syncing storage channel cache" in line:
            print(f"🔄 [CACHE SYNC] {line}")
        elif "Fast upload complete" in line or "Uploaded" in line:
            print(f"🚀 [UPLOAD COMPLETE] {line}")
        elif any(k in line.lower() for k in ["error", "fail", "exception", "floodwait"]):
            print(f"⚠️ [SYSTEM ALERT] {line}")
        else:
            print(f"   {line}")
        sys.stdout.flush()

    process.wait()
except KeyboardInterrupt:
    print("\n🛑 Stopping bot gracefully...")
    process.terminate()
    try:
        process.wait(timeout=5)
    except subprocess.TimeoutExpired:
        process.kill()
    print("✅ Bot stopped cleanly. Click Run (▶️) to restart anytime.")
```
